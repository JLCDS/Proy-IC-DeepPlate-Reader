# Dataset Exploration
Visualize samples, check class distribution, and audit annotation quality.

```python
import sys
sys.path.insert(0, '../src')

from pathlib import Path
import cv2
import matplotlib.pyplot as plt
import numpy as np

DATA_DIR = Path('../data/raw/plates/images/train')
```

```python
images = list(DATA_DIR.glob('*.jpg')) + list(DATA_DIR.glob('*.png'))
print(f'Total images: {len(images)}')

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
for ax, img_path in zip(axes.flat, images[:8]):
    img = cv2.cvtColor(cv2.imread(str(img_path)), cv2.COLOR_BGR2RGB)
    ax.imshow(img)
    ax.set_title(img_path.name, fontsize=8)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
