# Enhancement Pipeline Analysis
Compare raw vs enhanced plate crops side-by-side.

```python
import sys
sys.path.insert(0, '../src')

import cv2
import matplotlib.pyplot as plt
from deepplate.enhancement import ImageEnhancer
from deepplate.enhancement.preprocessor import EnhancerConfig

# Load a sample plate crop
img = cv2.imread('../data/samples/sample_plate.jpg')

configs = [
    ('No enhancement', EnhancerConfig(denoise=False, upscale_factor=1)),
    ('Upscale x2', EnhancerConfig(denoise=False, upscale_factor=2)),
    ('Full pipeline', EnhancerConfig()),
]

fig, axes = plt.subplots(1, len(configs), figsize=(16, 4))
for ax, (title, cfg) in zip(axes, configs):
    enhanced = ImageEnhancer(cfg).enhance(img)
    ax.imshow(cv2.cvtColor(enhanced, cv2.COLOR_BGR2RGB))
    ax.set_title(title)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
